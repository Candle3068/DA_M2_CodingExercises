Exercise Set 3: Duplicates

```python
data = {
"Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "Diana"],
"Age": [25, 30, 35, 25, 30, 40],
"City": ["NYC", "LA", "SF", "NYC", "LA", "NYC"],
}

df = pd.DataFrame(data)
```

```python
# 1 Find which rows are duplicates across all columns. COMPLETE

data = {
"Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "Diana"],
"Age": [25, 30, 35, 25, 30, 40],
"City": ["NYC", "LA", "SF", "NYC", "LA", "NYC"],
}

df = pd.DataFrame(data)

df.duplicated()
```

```python
# 2 Drop duplicate rows, keeping the first occurrence. COMPLETE

data = {
"Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "Diana"],
"Age": [25, 30, 35, 25, 30, 40],
"City": ["NYC", "LA", "SF", "NYC", "LA", "NYC"],
}

df = pd.DataFrame(data)

df = df.drop_duplicates()
```

```python
# 3 Drop duplicate rows, keeping only the last occurrence. COMPLETE

data = {
"Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "Diana"],
"Age": [25, 30, 35, 25, 30, 40],
"City": ["NYC", "LA", "SF", "NYC", "LA", "NYC"],
}

df = pd.DataFrame(data)

df = df.drop_duplicates(keep="last")
```

```python
# 4 Drop all duplicates, removing both the original and duplicate rows. COMPLETE

data = {
"Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "Diana"],
"Age": [25, 30, 35, 25, 30, 40],
"City": ["NYC", "LA", "SF", "NYC", "LA", "NYC"],
}

df = pd.DataFrame(data)
df.drop_duplicates(keep = False)
```

Exercise Set 1: Renaming Columns

```python
data = {
"fname": ["Alice", "Bob", "Charlie"],
"lname": ["Smith", "Jones", "Brown"],
"ctry": ["USA", "Canada", "UK"],
}

df = pd.DataFrame(data)
```

```python
# 1 Change fname to FirstName and lname to LastName. COMPLETE

data = {
"fname": ["Alice", "Bob", "Charlie"],
"lname": ["Smith", "Jones", "Brown"],
"ctry": ["USA", "Canada", "UK"],
}

df = pd.DataFrame(data)

df.rename(columns= {
    "fname" : "FirstName",
    "lname" : "LastName"
})
```

```python
# 2 Change all column names to: ['FirstName', 'LastName', 'Country'] COMPLETE

data = {
"fname": ["Alice", "Bob", "Charlie"],
"lname": ["Smith", "Jones", "Brown"],
"ctry": ["USA", "Canada", "UK"],
}

df = pd.DataFrame(data)
df.rename(columns= {
    "fname" : "FirstName",
    "lname" : "LastName",
    "ctry" : "Country"
})
```
